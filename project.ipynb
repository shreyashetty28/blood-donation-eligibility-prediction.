```python
def predict_blood_donation_eligibility(
        gender='Male',
        age=25,
        blood_group='O+',
        weight=65,
        hemoglobin=14):
    gender={'Female':0,'Male':1,'Other':2}[gender]
    blood_group = lb_blood.transform([blood_group])[0]

    data = [[gender,
             age,
             blood_group,
             weight,
             hemoglobin]]
    print(data)
    print(type(data))

    result = model.predict(data)


    if result[0] == 1:
        return "Person is Eligible for Blood Donation"

    else:
        return "Person is NOT Eligible for Blood Donation"

        
```

```python
print(predict_blood_donation_eligibility())
```

```python

gender = input("Enter Gender (Male/Female): ")

age = int(input("Enter Age: "))

blood_group = input("Enter Blood Group (A+, A-, B+, B-, AB+, AB-, O+, O-): ")

weight = float(input("Enter Weight (kg): "))

hemoglobin = float(input("Enter Hemoglobin (g/dL): "))

result = predict_blood_donation_eligibility(
            gender,
            age,
            blood_group,
            weight,
            hemoglobin)

print(result)
```
